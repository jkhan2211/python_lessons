## Calculate BMI

## Player's Position
Convert the categorical values in Player's position column in integer values

```python
df_20[['short_name', 'player_positions']]
```

```python
new_player_positions = df_20['player_positions'].str.get_dummies(sep= ', ').add_prefix('Position')
new_player_positions.head()
```

```python
df_20 = pd.concat([df_20, new_player_positions], axis= 1)
```

```python
df_20.head()
```

```python
# Lets drop the original player_positions columns
df_20 = df_20.drop('player_positions', axis=1)
```

```python
df_20.head()
```

```python
cols = ['ls','st','rs','lw','lf','cf','rf','rw','lam','cam','ram',
       'lm','lcm','cm','rcm','rm','lwb','ldm','cdm','rdm','rwb','lb',
       'lcb','cb','rcb','rb']
```

```python
df_20[cols].head()
```

```python
for col in cols:
    df_20[col] = df_20[col].str.split('+', n = 1, expand=True)[0]
df_20[cols]
```

```python
df_20[col] = df_20[col].fillna(0)
```

```python
df_20[col] = df_20[col].astype(int)
```

```python
df_20[cols]
```

## Filling Missing Values

```python
cols = ['dribbling','defending','physic','passing','shooting','pace']
df_20[cols]
```

```python
df_20[cols].isnull().sum()
```

```python
for col in cols:
    df_20[col] = df_20[col].fillna(df_20[col].median())
df_20[cols]
```
