## Task 2

### Data processing

Dropping some useless columns

```python
useless_col = ['dob','sofifa_id','player_url','long_name','body_type','real_face','loaned_from','nation_position','nation_jersey_number']
```

## Calculate BMI
lets calculate body max index of each player

```python
df_20['BMI'] = df_20['weight_kg'] / (df_20['height_cm']/100) ** 2
```

```python
df_20.head()
```

## Player's Position
Convert the categorical values in Player's position column in integer values

```python
df_20[['short_name', 'player_positions']]
```

```python
new_player_positions = df_20['player_positions'].str.get_dummies(sep= ', ').add_prefix('Position')
new_player_positions.head()
```

```python
df_20 = pd.concat([df_20, new_player_positions], axis= 1)
```

```python
df_20.head()
```

```python
# Lets drop the original player_positions columns
df_20 = df_20.drop('player_positions', axis=1)
```

```python
df_20.head()
```

```python
cols = ['ls','st','rs','lw','lf','cf','rf','rw','lam','cam','ram',
       'lm','lcm','cm','rcm','rm','lwb','ldm','cdm','rdm','rwb','lb',
       'lcb','cb','rcb','rb']
```

```python
df_20[cols].head()
```

```python
for col in cols:
    df_20[col] = df_20[col].str.split('+', n = 1, expand=True)[0]
df_20[cols]
```

```python
df_20[col] = df_20[col].fillna(0)
```

```python
df_20[col] = df_20[col].astype(int)
```

```python
df_20[cols]
```

## Filling Missing Values

```python
cols = ['dribbling','defending','physic','passing','shooting','pace']
df_20[cols]
```

```python
df_20[cols].isnull().sum()
```

```python
for col in cols:
    df_20[col] = df_20[col].fillna(df_20[col].median())
df_20[cols]
```

## Task 4: Exploratory Data Analysis 

1- Scatter Plot (colored by Age) year 2020 - Overall Rating vs Value in Errors


```python
fig = go.Figure(
    data= go.Scatter(
    x = df_20['overall'],
    y = df_20['value_eur'],
    mode = 'markers',
    marker = dict(
        size = 10,
        color = df_20['age'],
        ),
        text = df_20['short_name']

)
)
fig.update_layout(title='Scatter Plot (colored by age) year 2020 - overall ratings vs value in Eurors',
                  xaxis_title ='Overall Rating',
                  yaxis_title= 'Value in euros')
fig.show()
```
